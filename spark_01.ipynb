```python
data = [
    ("2026-11-01", "Wind", 12500),
    ("2026-11-01", "Solar", 4300),
    ("2026-11-01", "Gas", 8000),
    ("2026-11-02", "Wind", 14000),
]

columns = ["date", "source", "generation_mw"]

df = spark.createDataFrame(data, columns)

display(df)
```

```python
df.createOrReplaceTempView("energy_generation")
```

```python
%sql
SELECT source ,
    AVG(generation_mw) AS avg_generation_mw
FROM energy_generation
GROUP BY source 
ORDER BY avg_generation_mw DESC
```
