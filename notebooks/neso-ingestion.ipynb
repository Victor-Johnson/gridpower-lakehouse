```python
# importing pyspark functions into projects
from pyspark.sql import functions as F

file_path = "/Volumes/ledger_link/default/raw_landing/demanddataupdate.csv"

raw_df = (
    spark.read
    .option("header",True)
    .option("inferSchema",True)
    .csv(file_path)
)

display(raw_df.limit(20))
```

```python
raw_df.printSchema()
```

```python
print("Rows:", raw_df.count())
print("Columns:", len(raw_df.columns))

print("Column names:", raw_df.columns)
```

```python
bronze_df = (
    raw_df
    .withColumn("ingested_at", F.current_timestamp())
    .withColumn("source_system", F.lit("NESO"))
    .withColumn("source_file", F.lit(file_path))
)
```

```python
bronze_df.printSchema()
```

```python
from pyspark.sql import functions as F 

df = (
    spark.read
    .option("header", True)
    .csv(file_path)
    .select(
        "*",
        F.col("_metadata.file_path").alias("_source_file")
    )
)

df_bronze = df.withColumn(
    "_ingested_at",
    F.current_timestamp()
)
```

```python
df_bronze.printSchema()

display(
    df_bronze.select(
        "_ingested_at",
        "_source_file"
    ).limit(5)
)
```
